````python
from openai import OpenAI
from getpass import getpass

client = OpenAI(api_key=getpass("Enter your API key: "))
````

````python
def ask_agent(instructions, task):
    response = client.responses.create(
        model="gpt-5.6",
        instructions=instructions,
        input=task
    )
    return response.output_text
````

````python
def developer_agent(task):
    return ask_agent(
        "You are a software developer. Write simple, correct Python code for the user's task.",
        task
    )

def reviewer_agent(code):
    return ask_agent(
        "You are a code reviewer. Find bugs, errors, and improvements in this code.",
        code
    )

def tester_agent(code):
    return ask_agent(
        "You are a software tester. Create simple test cases for this code.",
        code
    )
````

````python
def autonomous_team(task):
    print("👨‍💻 DEVELOPER")
    code = developer_agent(task)
    print(code)

    print("\n" + "="*50)
    print("🔍 CODE REVIEWER")
    review = reviewer_agent(code)
    print(review)

    print("\n" + "="*50)
    print("🧪 TESTER")
    tests = tester_agent(code)
    print(tests)
````
